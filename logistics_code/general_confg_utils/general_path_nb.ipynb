```python
dbutils.widgets.text("project_cat","logistics_cat")
cat=dbutils.widgets.get("project_cat").strip()
dbutils.widgets.text("project_sch","logistics_sch")
sch=dbutils.widgets.get("project_sch").strip()
```

```python
%python
spark.sql(f"create catalog if not exists {cat}") 
```

```python
spark.sql(f"create schema if not exists {cat}.{sch}")
```

```python
spark.sql(f"create volume if not exists {cat}.{sch}.datalake")
spark.sql(f"create volume if not exists {cat}.{sch}.bronze")
spark.sql(f"create volume if not exists {cat}.{sch}.silver")
spark.sql(f"create volume if not exists {cat}.{sch}.gold")
```

```python
src=f"/Volumes/{cat}/{sch}/datalake"
brz=f"/Volumes/{cat}/{sch}/bronze"
slv=f"/Volumes/{cat}/{sch}/silver"
gld=f"/Volumes/{cat}/{sch}/gold"
slvdb=f"{cat}.{sch}"
glddb=f"{cat}.{sch}"
```

```python
import json
dbutils.notebook.exit(
json.dumbs({
    "CATALOG":cat,
    "SCHEMA":sch,
    "SOURCE":src,
    "BRONZE":brz,
    "SILVER":slv,
    "GOLD":gld,
    "SILVERDB":slvdb,
    "GOLDBDB":glddb
}))
```
